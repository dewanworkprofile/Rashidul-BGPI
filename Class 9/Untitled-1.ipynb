```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = sns.load_dataset('titanic')
```

```python
plt.figure(figsize=(10,9))
count=sns.countplot(data=df,x='sex',hue='survived')
for i in count.containers:
    count.bar_label(i)
    # count.bar_label(i)

plt.title('Quantity count')
plt.xlabel('Age')
plt.ylabel('Survived')
```

```python
plt.figure(figsize=(10,9))
count=sns.barplot(data=df,x='sex',y='survived',estimator='sum',color='red')
for i in count.containers:
    count.bar_label(i)
    # count.bar_label(i)

plt.title('Quantity count')
plt.xlabel('Age')
plt.ylabel('Survived')
```

```python
plt.figure(figsize=(10,9))
sns.boxplot(data=df,x='sex',y='age',hue='survived')
plt.title('Age Distribution by Sex and Survival')
plt.xlabel('Age',size=2)
plt.ylabel('Survived',size=2)
# plt.show()
```

```python
plt.figure(figsize=(10,9))
box=sns.boxplot(data=df,x='pclass',y='fare')
pclass_median=df[df['pclass']==1]['fare'].median()
box.annotate(
    'Median',
    xy=(0,pclass_median),
    xytext=(0.5,pclass_median+60),
    arrowprops={
        'arrowstyle':'->'
    }
)
pclass_outlier=df[df['pclass']==1]['fare'].max()
box.annotate(
    'outlier',
    xy=(0,pclass_outlier),
    xytext=(0.5,pclass_outlier-30),
    arrowprops={
        'arrowstyle':'->'
    }
)
```
