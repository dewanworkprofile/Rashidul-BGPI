```python
missing = df["rating"].str.contains("min" , na=False)
df[missing]
```

```python
df.loc[missing,"duration"] = df.loc[missing, "rating"]
```

```python
df["duration"].isnull().sum()
```

```python
df[df["rating"].isnull()]
```

```python
df.loc[missing,"rating"] = np.nan
```

```python
df["rating"].isnull().sum()
```

```python
df["rating"].unique()
```

```python
df["director"] = df["director"].fillna("Unknown")
```

```python
df["director"].isna().sum()
```

```python
df["cast"] = df["cast"].fillna("Unknown")
```

```python
df["cast"].isna().sum()
```

```python
df["country"] = df["country"].fillna("Unknown")
```

```python
df["country"].isna().sum()
```

```python
df["rating"] = df["rating"].fillna(df["rating"].mode()[0])
df["rating"].isna().sum()
```

```python
df['date_added'] = pd.to_datetime(df['date_added'].str.strip(), errors='coerce')
df['year_added'] = df['date_added'].dt.year
df[['date_added', 'year_added']].head()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
df.isnull().sum()
```

```python
plt.figure(figsize=(8, 5))
bar = sns.countplot(data=df, x='type', palette=["green","red"])
for container in bar.containers:
    bar.bar_label(container)
plt.title('Movies vs TV Shows on Netflix',size=20,color="blue")
plt.xlabel('Type',size=10, color="purple")
plt.ylabel('Count',size= 10  ,color="purple")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='rating', order=df['rating'].value_counts().index, palette='viridis')
plt.title('Content Rating Distribution',size=20,color="blue")
plt.xlabel('Rating',size=10,color="purple")
plt.ylabel('Count',size=10,color="purple")
plt.xticks(rotation=45)
plt.show()
```

```python
recent = df[(df['release_year'] >= 2015) & (df['release_year'] <= 2026)]
plt.figure(figsize=(12, 5))
sns.countplot(data=recent, x='release_year', hue='type', palette='Set1')
plt.title('Release Year Distribution (2015-2026): Movie vs TV Show',size=20,color="blue")
plt.xlabel('Release Year',size=10,color="purple")
plt.ylabel('Count',size=10,color="purple")
plt.legend()
plt.show()
```

```python
movies = df[df["type"]  == "Movie"]
countries = (movies["country"].str.split(",")
             .explode()
             .str.strip())

countries = countries[countries != "Unknown"]
top_countries = countries.value_counts().head(10)
top_countries
```

```python


plt.figure(figsize=(10, 5))
sns.barplot(x=top_countries.values, y=top_countries.index, palette='Set1')
plt.title('Top 10 Countries Producing the Most Movies',size=20,color="blue")
plt.xlabel('Number of Movies',size=10,color="purple")
plt.ylabel('Country',size=10,color="purple")
plt.show()
```

```python
genres = (movies["listed_in"].str.split(",")
          .explode()
          .str.strip())
top_genres = genres.value_counts().head(10)
top_genres
```

```python
plt.figure(figsize=(10, 5))
sns.barplot(x=top_genres.values, y=top_genres.index, palette='rocket')
plt.title('Top 10 Movie Categories on Netflix',size=20,color="blue")
plt.xlabel('Number of Movies',size=10,color="purple")
plt.ylabel('Category',size=10,color="purple")
plt.show()
```
