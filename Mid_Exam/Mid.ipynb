```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('netflix_titles_2021.csv')
display(df.head(15))
display(df.tail(15))
```

```python
print( df.shape)
print( list(df.columns))
```

```python
df.info()
```

```python
display(df.describe())
text_columns = df.select_dtypes(include='object')
display(text_columns.describe())
```

```python
missing_values = df.isnull().sum()
print(missing_values)
print(list(missing_values[missing_values > 0].index))
```

```python
duration_rows = df['rating'].str.contains('min', na=False)
print( duration_rows.sum())
```

```python
df.loc[duration_rows, 'duration'] = df.loc[duration_rows, 'rating']
df.loc[duration_rows, 'rating'] = np.nan
display(df.loc[duration_rows, ['title', 'rating', 'duration']])
```

```python
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')
```

```python
rating_mode = df['rating'].mode()[0]
df['rating'] = df['rating'].fillna(rating_mode)
print('Rating mode:', rating_mode)
```

```python
df['date_added'] = pd.to_datetime(df['date_added'].str.strip(), errors='coerce')
df['year_added'] = df['date_added'].dt.year
display(df[['added_date', 'added_year']].head())
```

```python
print( df.duplicated().sum())
df = df.drop_duplicates()
print(df.isnull().sum())
```

```python
plt.figure(figsize=(6, 5))
ax = sns.countplot(data=df, x='type', color='purple')
for i in ax.containers:
    ax.bar_label(i)
plt.title('Movies vs TV Shows on Netflix')
plt.xlabel('Type')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
rating_order = df['rating'].value_counts().index

plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='rating', order=rating_order, color='steelblue')
plt.title('Cnt Distribution')
plt.xlabel('Rating')
plt.ylabel('Cnt')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
recent = df[(df['release_year'] >= 2015) & (df['release_year'] <= 2026)]
year_order = list(range(2015, 2027))

plt.figure(figsize=(12, 5))
sns.countplot(data=recent, x='release_year', hue='type', order=year_order, palette='Set1')
plt.title('Release Year Distribution (2015–2026)')
plt.xlabel('Release Year')
plt.ylabel('Count')
plt.legend(title='Type')
plt.tight_layout()
plt.show()

print(pd.crosstab(recent['release_year'], recent['type']).reindex(year_order, fill_value=0))
```

```python
movies = df[df['type'] == 'Movie']

movie_countries = movies['country'].str.split(',').explode()
movie_countries = movie_countries.str.strip()
movie_countries = movie_countries[movie_countries != 'Unknown']
top_countries = movie_countries.value_counts().head(10)

print(top_countries)

plt.figure(figsize=(10, 5))
sns.barplot(x=top_countries.values, y=top_countries.index, color='steelblue')
plt.title('Top 10 Countries Producing the Most Movies')
plt.xlabel('Number of Movies')
plt.ylabel('Country')
plt.tight_layout()
plt.show()
```

```python
movie_categories = movies['listed_in'].str.split(',').explode()
movie_categories = movie_categories.str.strip()
top_categories = movie_categories.value_counts().head(10)

print(top_categories)

plt.figure(figsize=(10, 5))
sns.barplot(x=top_categories.values, y=top_categories.index, color='steelblue')
plt.title('Top 10 Movie Categories on Netflix')
plt.xlabel('Number of Movies')
plt.ylabel('Category')
plt.tight_layout()
plt.show()
```
