```python
missing = df['rating'].str.contains('min', na=False)
df.loc[missing, ['title', 'rating', 'duration']]
```

```python
df.loc[missing, 'duration'] = df.loc[missing, 'rating']
```

```python
df['duration'].isnull().sum()
```

```python
df[df['rating'].isnull()]
```

```python
df.loc[missing, 'rating'] = np.nan
```

```python
df['rating'].isnull().sum()
```

```python
df['rating'].unique()
```

```python
df['director'] = df['director'].fillna('Unknown')
```

```python
df['director'].isnull().sum()
```

```python
df['cast'] = df['cast'].fillna('Unknown')
```

```python
df['cast'].isnull().sum()
```

```python
df['country'] = df['country'].fillna('Unknown')
```

```python
df['country'].isnull().sum()
```

```python
rating_mode = df['rating'].mode()[0]
df['rating'] = df['rating'].fillna(rating_mode)
df['rating'].isnull().sum()
```

```python
df['date_added'] = pd.to_datetime(df['date_added'].str.strip(), errors='coerce')
df['year_added'] = df['date_added'].dt.year
df[['date_added', 'year_added']].head()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
df.isnull().sum()
```

```python
plt.figure(figsize=(8, 5))
bar = sns.countplot(
    data=df,
    x='type',
    hue='type',
    palette={'Movie': '#287C8E', 'TV Show': '#F28E6B'},
    legend=False
)
for container in bar.containers:
    bar.bar_label(container)
plt.title('Movies vs TV Shows on Netflix', color='#244653', fontsize=16)
plt.xlabel('Type', color='#52616B')
plt.ylabel('Count', color='#52616B')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
rating_order = df['rating'].value_counts().index
sns.countplot(
    data=df,
    x='rating',
    order=rating_order,
    color='#8C75B8'
)
plt.title('Content Rating Distribution', color='#244653', fontsize=16)
plt.xlabel('Rating', color='#52616B')
plt.ylabel('Count', color='#52616B')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
recent = df[(df['release_year'] >= 2015) & (df['release_year'] <= 2026)]
plt.figure(figsize=(12, 5))
sns.countplot(
    data=recent,
    x='release_year',
    hue='type',
    order=list(range(2015, 2027)),
    palette={'Movie': '#287C8E', 'TV Show': '#F4B942'}
)
plt.title('Release Year Distribution (2015-2026)', color='#244653', fontsize=16)
plt.xlabel('Release Year', color='#52616B')
plt.ylabel('Count', color='#52616B')
plt.legend(title='Type')
plt.tight_layout()
plt.show()
```

```python
movies = df[df['type'] == 'Movie']
countries = movies['country'].str.split(',').explode()
countries = countries.str.strip()
countries = countries[countries != 'Unknown']
top_countries = countries.value_counts().head(10)
top_countries
```

```python
plt.figure(figsize=(10, 5))
sns.barplot(
    x=top_countries.values,
    y=top_countries.index,
    color='#E69A65'
)
plt.title('Top 10 Countries Producing the Most Movies', color='#244653', fontsize=16)
plt.xlabel('Number of Movies', color='#52616B')
plt.ylabel('Country', color='#52616B')
plt.tight_layout()
plt.show()
```

```python
categories = movies['listed_in'].str.split(',').explode()
categories = categories.str.strip()
top_categories = categories.value_counts().head(10)
top_categories
```

```python
plt.figure(figsize=(10, 5))
sns.barplot(
    x=top_categories.values,
    y=top_categories.index,
    color='#4F9D8F'
)
plt.title('Top 10 Movie Categories on Netflix', color='#244653', fontsize=16)
plt.xlabel('Number of Movies', color='#52616B')
plt.ylabel('Category', color='#52616B')
plt.tight_layout()
plt.show()
```
